### Text Detection

Kiểm tra dữ liệu từ file XML

```python
import os
import xml.etree.ElementTree as ET

def extract_data_from_xml(root_dir):
    xml_path = os.path.join(root_dir, 'words.xml')
    tree = ET.parse(xml_path)
    root = tree.getroot()

    img_paths = []
    img_sizes = []
    img_labels = []
    bboxes = []

    for img in root:
        bbs_of_img = []
        labels_of_img = []

        tagged_rectangles = img.find('taggedRectangles')
        if tagged_rectangles is None:
            continue  # bỏ qua nếu không có bbox

        for bb in tagged_rectangles:
            text_node = bb.find('tag')
            if text_node is None:
                continue

            text = text_node.text
            if text is None or not text.isalnum():
                continue  # bỏ nhãn không phải chữ/số

            text = text.lower()
            if ' ' in text:
                continue  # bỏ nhãn có khoảng trắng

            try:
                x = float(bb.attrib['x'])
                y = float(bb.attrib['y'])
                w = float(bb.attrib['width'])
                h = float(bb.attrib['height'])
            except KeyError:
                continue  # bỏ nếu thiếu thông tin bbox

            bbs_of_img.append([x, y, w, h])
            labels_of_img.append(text)

        # Bỏ qua ảnh không có bbox hợp lệ
        if not bbs_of_img:
            continue

        img_path = os.path.join(root_dir, img[0].text)
        width = int(img[1].attrib['x'])
        height = int(img[1].attrib['y'])

        img_paths.append(img_path)
        img_sizes.append((width, height))
        bboxes.append(bbs_of_img)
        img_labels.append(labels_of_img)

    return img_paths, img_sizes, img_labels, bboxes


# Example usage
dataset_dir = '/content/drive/MyDrive/OCR : Yolov11 vs CNN/SceneTrialTrain'
img_paths, img_sizes, img_labels, bboxes = extract_data_from_xml(dataset_dir)
print(img_paths[1])
print(img_sizes[1])
print(img_labels[1])
print(bboxes[1])
```

Chuyển sang bbox định dạng data của Yolo

```python
def convert_to_yolo_format(image_paths, image_sizes, bounding_boxes):
    """
    Chuyển đổi thông tin bbox sang YOLO format:
    - Định dạng: (class_id, x_center, y_center, width, height) – chuẩn hóa [0, 1]
    - Mỗi ảnh gắn kèm danh sách các dòng nhãn
    """
    yolo_data = []

    for image_path, image_size, bboxes in zip(image_paths, image_sizes, bounding_boxes):
        img_w, img_h = image_size
        yolo_labels = []

        for bbox in bboxes:
            x, y, w, h = bbox

            # Tọa độ tâm + kích thước chuẩn hóa
            x_center = (x + w / 2) / img_w
            y_center = (y + h / 2) / img_h
            w_norm = w / img_w
            h_norm = h / img_h

            class_id = 0  # vì chỉ có 1 lớp: "text"

            yolo_label = f"{class_id} {x_center:.6f} {y_center:.6f} {w_norm:.6f} {h_norm:.6f}"
            yolo_labels.append(yolo_label)

        # Gắn ảnh với danh sách nhãn tương ứng
        yolo_data.append((image_path, yolo_labels))

    return yolo_data

class_labels = ['text']
yolo_data = convert_to_yolo_format(img_paths, img_sizes, bboxes)
```

### Text Recognition

```python
import os
import random
import time
import xml . etree . ElementTree as ET

import cv2
import matplotlib . pyplot as plt
import numpy as np
import timm
import torch
import torch . nn as nn
import torchvision
from PIL import Image
from sklearn . model_selection import train_test_split
from torch . nn import functional as F
from torch . utils . data import DataLoader , Dataset
from torchvision import transforms
```

Cắt ảnh từ bboxs tạo dataset cho CRNN

```python
def split_bounding_boxes(img_paths, img_labels, bboxes, save_dir):
    os.makedirs(save_dir, exist_ok=True)

    count = 0
    labels = []

    for img_path, label_list, bbox_list in zip(img_paths, img_labels, bboxes):
        try:
            img = Image.open(img_path).convert("RGB")  # chuyển về RGB nếu ảnh xám
        except Exception as e:
            print(f"⚠️ Không thể mở ảnh: {img_path} ({e})")
            continue

        for label, box in zip(label_list, bbox_list):
            x, y, w, h = map(int, box)
            cropped_img = img.crop((x, y, x + w, y + h))

            # Convert to numpy để tính trung bình pixel
            img_np = np.array(cropped_img)
            img_mean = img_np.mean()

            # Bỏ qua nếu ảnh quá tối/sáng hoặc quá nhỏ
            if img_mean < 35 or img_mean > 220:
                continue
            if cropped_img.width < 10 or cropped_img.height < 10:
                continue

            # Lưu ảnh
            filename = f"{count:06d}.jpg"
            save_path = os.path.join(save_dir, filename)
            cropped_img.save(save_path)

            # Ghi nhãn: đường dẫn + chuỗi text
            labels.append(f"{save_path}\t{label}")
            count += 1

    print(f"✅ Đã tạo {count} ảnh đã crop")

    # Ghi toàn bộ nhãn ra file labels.txt
    label_file = os.path.join(save_dir, "labels.txt")
    with open(label_file, "w") as f:
        for line in labels:
            f.write(line + "\n")

save_dir = "/content/drive/MyDrive/OCR : Yolov11 vs CNN/ocr_dataset"
split_bounding_boxes(img_paths, img_labels, bboxes, save_dir)
```

Đọc dữ liệu

```python
root_dir = save_dir

img_paths = []
labels = []

# Read labels from text file
with open(os.path.join(root_dir, "labels.txt"), "r") as f:
    for label in f:
        parts = label.strip().split("\t")
        if len(parts) == 2:
            img_paths.append(parts[0])
            labels.append(parts[1])
        else:
            # Handle cases where a line might not have the expected format
            print(f"Skipping malformed line: {label.strip()}")

print(f"Total images: {len(img_paths)}")
```

Xậy dựng bộ từ vựng

```python
letters = [char.split(".")[0].lower() for char in labels]
letters = "".join(letters)
letters = sorted(list(set(list(letters))))

# create a string of all characters in the dataset
chars = "".join(letters)

# for "blank" character
blank_char = "-"
chars += blank_char
vocab_size = len(chars)

print(f"Vocab: {chars}")
print(f"Vocab size: {vocab_size}")
```

```python
char_to_idx = {char: idx + 1 for idx, char in enumerate (sorted(chars))}
idx_to_char = {index: char for char, index in char_to_idx.items()}
```

Encode

```python
max_label_len = max([len(label) for label in labels])

def encode(label, char_to_idx, max_label_len):
    encoded_labels = torch.tensor(
        [char_to_idx[char] for char in label],
        dtype=torch.int32
    )
    label_len = len(encoded_labels)
    lengths = torch.tensor(
        label_len,
        dtype=torch.int32
    )
    padded_labels = F.pad(
        encoded_labels,
        (0, max_label_len - label_len),
        value=0
    )

    return padded_labels, lengths
```

Decode

```python
def decode(encoded_sequences, idx_to_char, blank_char="-"):
    decoded_sequences = []

    for seq in encoded_sequences:
        decoded_label = []
        prev_char = None  # To track the previous character

        for token in seq:
            if token != 0:  # Ignore padding (token = 0)
                char = idx_to_char[token.item()]
                # Append the character if it's not a blank or the same as the previous character
                if char != blank_char:
                    if char != prev_char or prev_char == blank_char:
                        decoded_label.append(char)
                    prev_char = char  # Update previous character

        decoded_sequences.append("".join(decoded_label))

    print(f"From {encoded_sequences} to {decoded_sequences}")

    return decoded_sequences
```

Tiền xử lý dữ liệu

```python
data_transforms = {
    "train": transforms.Compose(
        [
            transforms.Resize((100, 420)),
            transforms.ColorJitter(
                brightness=0.5,
                contrast=0.5,
                saturation=0.5,
            ),
            transforms.Grayscale(
                num_output_channels=1,
            ),
            transforms.GaussianBlur(3),
            transforms.RandomAffine(
                degrees=1,
                shear=1,
            ),
            transforms.RandomPerspective(
                distortion_scale=0.3,
                p=0.5,
                interpolation=3,
            ),
            transforms.RandomRotation(degrees=2),
            transforms.ToTensor(),
            transforms.Normalize((0.5,), (0.5,)),
        ]
    ),
    "val": transforms.Compose(
        [
            transforms.Resize((100, 420)),
            transforms.Grayscale(num_output_channels=1),
            transforms.ToTensor(),
            transforms.Normalize((0.5,), (0.5,)),
        ]
    ),
}
```

Chia bộ dữ liệu

```python
seed = 0
val_size = 0.1
test_size = 0.1
is_shuffle = True

X_train, X_val, y_train, y_val = train_test_split(
    img_paths,
    labels,
    test_size=val_size,
    random_state=seed,
    shuffle=is_shuffle,
)

X_train, X_test, y_train, y_test = train_test_split(
    X_train,
    y_train,
    test_size=test_size,
    random_state=seed,
    shuffle=is_shuffle,
)
```

Định nghĩa lớp cho dataset

```python
class STRDataset(Dataset):
    def __init__(
        self,
        X,
        y,
        char_to_idx,
        max_label_len,
        label_encoder=None,
        transform=None,
    ):
        self.transform = transform
        self.img_paths = X
        self.labels = y
        self.char_to_idx = char_to_idx
        self.max_label_len = max_label_len
        self.label_encoder = label_encoder

    def __len__(self):
        return len(self.img_paths)

    def __getitem__(self, idx):
        label = self.labels[idx]
        img_path = self.img_paths[idx]
        img = Image.open(img_path).convert("RGB")

        if self.transform:
            img = self.transform(img)

        if self.label_encoder:
            encoded_label, label_len = self.label_encoder(
                label, self.char_to_idx, self.max_label_len
            )
            return img, encoded_label, label_len
```

Loader

```python
train_dataset = STRDataset(
    X_train,
    y_train,
    char_to_idx=char_to_idx,
    max_label_len=max_label_len,
    label_encoder=encode,
    transform=data_transforms["train"],
)
val_dataset = STRDataset(
    X_val,
    y_val,
    char_to_idx=char_to_idx,
    max_label_len=max_label_len,
    label_encoder=encode,
    transform=data_transforms["val"],
)
test_dataset = STRDataset(
    X_test,
    y_test,
    char_to_idx=char_to_idx,
    max_label_len=max_label_len,
    label_encoder=encode,
    transform=data_transforms["val"],
)

# Reduce batch sizes to mitigate OutOfMemoryError
train_batch_size = 64
test_batch_size = 64 * 2

train_loader = DataLoader(
    train_dataset,
    batch_size=train_batch_size,
    shuffle=True,
)
val_loader = DataLoader(
    val_dataset,
    batch_size=test_batch_size,
    shuffle=False,
)
test_loader = DataLoader(
    test_dataset,
    batch_size=test_batch_size,
    shuffle=False,
)
```

Model CRNN

```python
class CRNN(nn.Module):
    def __init__(
        self, vocab_size, hidden_size, n_layers, dropout=0.2, unfreeze_layers=3
    ):
        super(CRNN, self).__init__()

        backbone = timm.create_model("resnet152", in_chans=1, pretrained=True)
        modules = list(backbone.children())[:-2]
        modules.append(nn.AdaptiveAvgPool2d((1, None)))
        self.backbone = nn.Sequential(*modules)

        # Unfreeze the last few layers
        for parameter in self.backbone[-unfreeze_layers:].parameters():
            parameter.requires_grad = True

        self.mapSeq = nn.Sequential(
            nn.Linear(2048, 512), nn.ReLU(), nn.Dropout(dropout)
        )

        self.gru = nn.GRU(
            512,
            hidden_size,
            n_layers,
            bidirectional=True,
            batch_first=True,
            dropout=dropout if n_layers > 1 else 0,
        )
        self.layer_norm = nn.LayerNorm(hidden_size * 2)

        self.out = nn.Sequential(
            nn.Linear(hidden_size * 2, vocab_size), nn.LogSoftmax(dim=2)
        )

    @torch.autocast(device_type="cuda")
    def forward(self, x):
        x = self.backbone(x)
        x = x.permute(0, 3, 1, 2)
        x = x.view(x.size(0), x.size(1), -1)  # Flatten the feature map
        x = self.mapSeq(x)
        x, _ = self.gru(x)
        x = self.layer_norm(x)
        x = self.out(x)
        x = x.permute(1, 0, 2)  # Based on CTC

        return x
```

```python
hidden_size = 256
n_layers = 3
dropout_prob = 0.2
unfreeze_layers = 3
device = "cuda" if torch.cuda.is_available() else "cpu"

# Clear CUDA cache and free up memory
import gc
import torch

gc.collect()
torch.cuda.empty_cache()

model = CRNN(
    vocab_size=vocab_size,
    hidden_size=hidden_size,
    n_layers=n_layers,
    dropout=dropout_prob,
    unfreeze_layers=unfreeze_layers,
).to(device)
```

Train, Evaluate

```python
from torch.amp import GradScaler, autocast

def evaluate(model, dataloader, criterion, device):
    model.eval()
    losses = []
    with torch.no_grad():
        for inputs, targets, target_lens in dataloader:
            inputs      = inputs.to(device)
            targets     = targets.to(device)
            target_lens = target_lens.to(device)

            out = model(inputs)  # có thể trả (B, T, C) hoặc (T, B, C)

            # Nếu out.shape = (B, T, C), permute về (T, B, C)
            if out.dim()==3 and out.shape[0] == targets.size(0):
                out = out.permute(1, 0, 2)

            # Bây giờ chắc chắn out.shape = (T, B, C)
            T, B, _ = out.size()
            logits_lens = torch.full((B,), T, dtype=torch.long, device=device)

            loss = criterion(out, targets, logits_lens, target_lens)
            losses.append(loss.item())

    return sum(losses) / len(losses)


def fit(model, train_loader, val_loader,
        criterion, optimizer, scheduler,
        device, epochs,
        use_amp=True, max_grad_norm=2.0):
    scaler = GradScaler() if use_amp else None
    train_losses, val_losses = [], []

    for epoch in range(1, epochs+1):
        model.train()
        batch_losses = []
        t0 = time.time()

        for inputs, targets, target_lens in train_loader:
            inputs      = inputs.to(device)
            targets     = targets.to(device)
            target_lens = target_lens.to(device)

            optimizer.zero_grad()

            if use_amp:
                with autocast(device_type="cuda"):
                    out = model(inputs)
                    if out.dim()==3 and out.shape[0] == targets.size(0):
                        out = out.permute(1, 0, 2)
                    T, B, _ = out.size()
                    logits_lens = torch.full((B,), T, dtype=torch.long, device=device)
                    loss = criterion(out, targets, logits_lens, target_lens)

                scaler.scale(loss).backward()
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_grad_norm)
                scaler.step(optimizer)
                scaler.update()
            else:
                out = model(inputs)
                if out.dim()==3 and out.shape[0] == targets.size(0):
                    out = out.permute(1, 0, 2)
                T, B, _ = out.size()
                logits_lens = torch.full((B,), T, dtype=torch.long, device=device)
                loss = criterion(out, targets, logits_lens, target_lens)

                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_grad_norm)
                optimizer.step()

            batch_losses.append(loss.item())

        train_loss = sum(batch_losses) / len(batch_losses)
        val_loss   = evaluate(model, val_loader, criterion, device)
        scheduler.step()

        elapsed = time.time() - t0
        print(f"Epoch {epoch}/{epochs} — "
              f"Train loss: {train_loss:.4f} — "
              f"Val loss: {val_loss:.4f} — "
              f"{elapsed:.1f}s")

        train_losses.append(train_loss)
        val_losses.append(val_loss)

    return train_losses, val_losses
```

Loss, optimizer, schedule

```python
epochs = 100
lr = 1e-5
weight_decay = 1e-5
scheduler_step_size = 20

criterion = nn.CTCLoss(
    blank=char_to_idx[blank_char],
    zero_infinity=True,
    reduction="mean",
)
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=lr,
    weight_decay=weight_decay,
)
scheduler = torch.optim.lr_scheduler.StepLR(
    optimizer, step_size=scheduler_step_size, gamma=0.5
)
```

```python
train_losses, val_losses = fit(
    model, train_loader, val_loader, criterion, optimizer, scheduler,
    device, epochs
)
```

```python
val_loss = evaluate(model, val_loader, criterion, device)
test_loss = evaluate(model, test_loader, criterion, device)

print("Evaluation on val/test dataset")
print("Val loss:", val_loss)
print("Test loss:", test_loss)
```
