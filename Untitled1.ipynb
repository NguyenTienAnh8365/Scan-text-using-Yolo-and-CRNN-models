```python
import torch
import torch.nn as nn
import timm
from collections import OrderedDict

class CRNN(nn.Module):
    def __init__(self, vocab_size, hidden_size, n_layers, dropout=0.2, unfreeze_layers=3):
        super(CRNN, self).__init__()
        # Backbone: ResNet101 pretrained on ImageNet, single-channel input
        backbone = timm.create_model('resnet101', in_chans=1, pretrained=True)
        modules = list(backbone.children())[:-2]
        modules.append(nn.AdaptiveAvgPool2d((1, None)))
        self.backbone = nn.Sequential(*modules)

        # Unfreeze last few layers of backbone
        for param in list(self.backbone.parameters())[-unfreeze_layers:]:
            param.requires_grad = True

        # Sequence mapping
        self.mapSeq = nn.Sequential(
            nn.Linear(2048, 512),
            nn.ReLU(),
            nn.Dropout(dropout)
        )

        # Bi-directional LSTM
        self.lstm = nn.LSTM(
            input_size=512,
            hidden_size=hidden_size,
            num_layers=n_layers,
            bidirectional=True,
            batch_first=True,
            dropout=dropout if n_layers > 1 else 0
        )

        self.layer_norm = nn.LayerNorm(hidden_size * 2)
        self.out = nn.Sequential(
            nn.Linear(hidden_size * 2, vocab_size),
            nn.LogSoftmax(dim=2)
        )

    def forward(self, x):
        # x: (B, 1, H, W)
        x = self.backbone(x)                  # (B, C=2048, 1, W')
        x = x.permute(0, 3, 1, 2)             # (B, W', C, 1)
        x = x.view(x.size(0), x.size(1), -1)  # (B, W', 2048)
        x = self.mapSeq(x)                    # (B, W', 512)
        x, _ = self.lstm(x)                   # (B, W', hidden*2)
        x = self.layer_norm(x)
        x = self.out(x)                       # (B, W', vocab_size)
        x = x.permute(1, 0, 2)                # (T=W', B, vocab_size) for CTC
        return x

# --------- Loading pretrained weights safely ---------
def load_crnn_weights(model, checkpoint_path, device):
    # Load checkpoint
    checkpoint = torch.load(checkpoint_path, map_location=device)
    state_dict = checkpoint if isinstance(checkpoint, dict) and 'state_dict' not in checkpoint else checkpoint['state_dict']

    # Filter out mismatched keys
    model_dict = model.state_dict()
    filtered_dict = OrderedDict()
    for k, v in state_dict.items():
        if k in model_dict and v.size() == model_dict[k].size():
            filtered_dict[k] = v
        else:
            print(f"Skipping key '{k}': saved shape {tuple(v.size())}, model shape {tuple(model_dict.get(k, v).size())}")

    # Load filtered weights
    model.load_state_dict(filtered_dict, strict=False)
    print("Loaded pretrained weights with mismatches skipped.")

# --------- Example initialization & load ---------
if __name__ == '__main__':
    # Parameters
    chars = '0123456789 abcdefghijklmnopqrstuvwxyz -'
    vocab_size = len(chars) + 1  # +1 if using index 0 as blank for CTC
    hidden_size = 256
    n_layers = 3
    dropout = 0.2
    unfreeze_layers = 3

    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    model_path = '/content/drive/MyDrive/OCR : Yolov11 vs CNN/ocr_crnn.pt'

    # Initialize model
    crnn = CRNN(
        vocab_size=vocab_size,
        hidden_size=hidden_size,
        n_layers=n_layers,
        dropout=dropout,
        unfreeze_layers=unfreeze_layers
    ).to(device)

    # Load weights
    load_crnn_weights(crnn, model_path, device)

    # Model ready for training or inference
    print("CRNN model is ready on device:", device)
```

```python
def decode(encoded_sequences, idx_to_char, blank_char='-'):
    """
    Hàm giải mã các chuỗi số (index) thành chuỗi ký tự.

    Args:
        encoded_sequences (List[Tensor]): list các tensor output từ model, đã qua argmax.
        idx_to_char (Dict[int, str]): ánh xạ từ chỉ số sang ký tự.
        blank_char (str): ký tự đại diện cho "blank" trong CTC.

    Returns:
        List[str]: danh sách chuỗi kết quả đã giải mã.
    """
    decoded_sequences = []
    blank_index = 1 # Assuming 1 is the blank index based on vocab_size = len(chars) + 1

    for seq in encoded_sequences:
        decoded_label = []
        prev_char = None  # để loại trừ ký tự lặp lại liên tục

        for token in seq:
            token_item = token.item()
            if token_item != 0:  # 0 thường là padding
                if token_item != blank_index: # Check if it's not the blank index
                    char = idx_to_char[token_item]
                    if char != prev_char:  # loại bỏ lặp liên tiếp
                        decoded_label.append(char)
                    prev_char = char
                else: # If it's the blank index
                    prev_char = None  # reset
            else:
                prev_char = None  # reset nếu là padding

        decoded_sequences.append("".join(decoded_label))

    return decoded_sequences
```

```python
def text_detection(img_path, text_det_model):
    """
    Phát hiện văn bản trong ảnh sử dụng mô hình YOLOv11.

    Args:
        img_path (str or PIL.Image): đường dẫn hoặc ảnh đầu vào.
        text_det_model: mô hình YOLOv11 đã huấn luyện.

    Returns:
        Tuple:
            - bboxes: List[List[float]]: danh sách bounding boxes [x1, y1, x2, y2]
            - classes: List[int]: chỉ số lớp
            - names: Dict[int, str]: ánh xạ class index -> class name
            - confs: List[float]: độ tin cậy của mỗi dự đoán
    """
    results = text_det_model(img_path, verbose=False)[0]

    bboxes = results.boxes.xyxy.tolist()     # Bounding box dạng [x1, y1, x2, y2]
    classes = results.boxes.cls.tolist()     # Chỉ số lớp
    names = results.names                    # Dict ánh xạ index -> tên lớp
    confs = results.boxes.conf.tolist()      # Confidence

    return bboxes, classes, names, confs
```

```python
def text_recognition(img, data_transforms, text_reg_model, idx_to_char, device):
    """
    Hàm nhận diện văn bản trong một ảnh đã được crop (từ bounding box).

    Args:
        img (PIL.Image or np.ndarray): ảnh chứa văn bản cần nhận diện.
        data_transforms (torchvision.transforms): phép biến đổi đầu vào cho CRNN.
        text_reg_model (nn.Module): mô hình nhận diện văn bản đã huấn luyện (CRNN).
        idx_to_char (dict): ánh xạ từ index -> ký tự.
        device (str): thiết bị thực thi ('cuda' hoặc 'cpu').

    Returns:
        str: chuỗi văn bản được nhận diện từ ảnh.
    """
    # Tiền xử lý ảnh (resize, normalize, grayscale, ...)
    transformed_image = data_transforms(img)                       # Tensor [C, H, W]
    transformed_image = transformed_image.unsqueeze(0).to(device) # Add batch -> [1, C, H, W]

    text_reg_model.eval()
    with torch.no_grad():
        # Dự đoán
        logits = text_reg_model(transformed_image)                 # [T, B=1, Vocab]
        logits = logits.detach().cpu()

        # Giải mã: lấy argmax theo chiều vocab và decode thành chuỗi
        pred_indices = logits.permute(1, 0, 2).argmax(2)           # [B, T]
        texts = decode(pred_indices, idx_to_char)

    return texts[0]  # Vì chỉ xử lý 1 ảnh, lấy phần tử đầu tiên
```

```python
def visualize_detections(img, detections):
    """
    Hiển thị ảnh và overlay kết quả phát hiện + nhận diện văn bản.

    Args:
        img (PIL.Image or np.ndarray): ảnh gốc.
        detections (List[Tuple[bbox, class_name, confidence, text]]):
            danh sách các phát hiện, mỗi phần tử gồm:
                - bbox: [x1, y1, x2, y2]
                - class_name: tên lớp (thường là 'text' hoặc id lớp)
                - confidence: độ tin cậy của YOLO
                - transcribed_text: kết quả nhận diện văn bản từ CRNN
    """
    plt.figure(figsize=(12, 8))
    plt.imshow(img)
    plt.axis("off")

    for bbox, detected_class, confidence, transcribed_text in detections:
        x1, y1, x2, y2 = bbox

        # Vẽ bounding box
        plt.gca().add_patch(
            plt.Rectangle(
                (x1, y1),
                x2 - x1,
                y2 - y1,
                fill=False,
                edgecolor='red',
                linewidth=2
            )
        )

        # Ghi chú lớp + độ tin cậy + văn bản
        plt.text(
            x1,
            y1 - 10,
            f"{detected_class} ({confidence:.2f}): {transcribed_text}",
            fontsize=9,
            bbox=dict(facecolor='red', alpha=0.5),
            color='white'
        )

    plt.show()
```

```python
data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((100, 420)),  # Resize ảnh về đúng shape đầu vào CRNN
        transforms.ColorJitter(
            brightness=0.5,
            contrast=0.5,
            saturation=0.5
        ),  # Augment màu sắc
        transforms.Grayscale(num_output_channels=1),  # Chuyển sang ảnh xám
        transforms.GaussianBlur(3),  # Làm mờ nhẹ
        transforms.RandomAffine(degrees=1, shear=1),  # Biến dạng nhẹ affine
        transforms.RandomPerspective(
            distortion_scale=0.2,
            p=0.3,
            interpolation=3
        ),  # Biến dạng phối cảnh
        transforms.RandomRotation(degrees=2),  # Xoay nhẹ
        transforms.ToTensor(),  # Chuyển ảnh về Tensor [C, H, W]
        transforms.Normalize((0.5,), (0.5,))  # Chuẩn hóa về mean=0, std=1
    ]),

    'val': transforms.Compose([
        transforms.Resize((100, 420)),
        transforms.Grayscale(num_output_channels=1),
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ]),
}
```

```python
def predict(img_path, data_transforms, text_det_model, text_reg_model,
            idx_to_char, device):
    """
    Pipeline phát hiện và nhận diện văn bản trong ảnh.

    Args:
        img_path (str): đường dẫn đến ảnh đầu vào.
        data_transforms: transform áp dụng cho CRNN.
        text_det_model: mô hình phát hiện văn bản (YOLO).
        text_reg_model: mô hình nhận diện văn bản (CRNN).
        idx_to_char (dict): ánh xạ index -> ký tự.
        device (str): 'cuda' hoặc 'cpu'.

    Returns:
        List[Tuple[bbox, class_name, confidence, transcribed_text]]
    """
    # Dùng mô hình YOLO phát hiện vùng chữ
    bboxes, classes, names, confs = text_detection(img_path, text_det_model)

    # Mở ảnh
    img = Image.open(img_path).convert("RGB")
    predictions = []

    for bbox, cls, conf in zip(bboxes, classes, confs):
        x1, y1, x2, y2 = map(int, bbox)
        name = names[int(cls)]

        # Crop vùng chữ và nhận diện văn bản
        cropped_image = img.crop((x1, y1, x2, y2))
        transcribed_text = text_recognition(
            cropped_image,
            data_transforms,
            text_reg_model,
            idx_to_char,
            device
        )

        predictions.append((bbox, name, conf, transcribed_text))

    # Hiển thị kết quả
    visualize_detections(img, predictions)

    return predictions
```
